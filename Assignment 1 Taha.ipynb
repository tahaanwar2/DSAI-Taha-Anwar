Q1. Write a program that takes two numbers (a = 15 , b = 4) and prints their:


- Sum

- Difference

- Product

- Division

- Floor Division

- Modulus

```python
a = 15
b = 4

print("Sum:", a + b)
print("Difference:", a - b)
print("Product:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
```

Q2. Check the type and convert:
(num1 = 10, num2 = 5.6, num3 = 4.43, num4='23')

- Integer to float

- Float to integer

- Float to String

- String to Integer

```python
num1 = 10
num2 = 5.6
num3 = 4.43
num4 = '23'

print("Original Types:")
print(type(num1))
print(type(num2))
print(type(num3))
print(type(num4))


num1 = float(num1)     
num2 = int(num2)       
num3 = str(num3)        
num4 = int(num4)        


print("\nConverted Values and Types:")
print(num1, type(num1))
print(num2, type(num2))
print(num3, type(num3))
print(num4, type(num4))
```

Q15.
Given the list:
data = [12, 45, 2, 67, 34, 89, 23, 90]

Perform the following tasks:

Sort the list in ascending order.
Remove the highest value from the list.
Insert the value 50 at index 3.".
Finally, reverse the updated list and display the result.

```python
data = [12, 45, 2, 67, 34, 89, 23, 90]
data.sort()

data.remove(max(data))

data.insert(3, 50)

data.reverse()

print(data)
```
